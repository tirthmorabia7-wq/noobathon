```python
import matplotlib.pyplot as plt
import numpy as np

xpts = np.array([1, 2, 3, 4, 5])
ypts = np.array([2, 4, 6, 8, 10])

plt.plot(xpts, ypts)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Plot')
plt.show()
```
